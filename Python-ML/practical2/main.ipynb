# Principal Component Analysis (PCA) Example

In this notebook, we will perform PCA to reduce the dimensionality of a dataset while retaining the maximum variance.  
We will create a simple synthetic dataset and visualize the transformed features in a lower-dimensional space.


## Step 1: Import Required Libraries


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Step 2: Create a Simple Dataset

We will create a 2D dataset with some correlation between features.


```python
# Create a simple dataset
data = {
    'Feature_1': [2.5, 0.5, 2.2, 1.9, 3.1, 2.3, 2, 1, 1.5, 1.1],
    'Feature_2': [2.4, 0.7, 2.9, 2.2, 3, 2.7, 1.6, 1.1, 1.6, 0.9]
}

df = pd.DataFrame(data)
df.head()
```

## Step 3: Standardize the Data

PCA works better when data is standardized (mean=0, variance=1).


```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
scaled_df.head()
```

## Step 4: Apply PCA

We will reduce the data from 2 dimensions to 1 dimension (for visualization purposes).


```python
# Apply PCA
pca = PCA(n_components=1)
pca_result = pca.fit_transform(scaled_df)

# Convert to DataFrame
pca_df = pd.DataFrame(pca_result, columns=['PCA_1'])
pca_df.head()
```

## Step 5: Explained Variance

Check how much variance is retained by the principal component.


```python
explained_variance = pca.explained_variance_ratio_
print(f"Explained variance by the first principal component: {explained_variance[0]:.2f}")
```

## Step 6: Visualize the Transformed Features

Since we reduced to 1D, we can plot the data along the principal component axis.


```python
plt.figure(figsize=(8, 4))
plt.scatter(pca_df['PCA_1'], np.zeros_like(pca_df['PCA_1']), color='blue')
plt.xlabel('Principal Component 1')
plt.title('PCA Transformation')
plt.legend(['Data points'])
plt.show()
```

✅ Summary

PCA reduces dimensionality while retaining maximum variance.

We created a simple 2D dataset and reduced it to 1D.

Explained variance tells us how much information is preserved.

Visualization shows the data projected along the principal component.
