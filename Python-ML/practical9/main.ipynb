2. Import Libraries

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

3. Create Dataset

```python
transactions = [
    ["Milk", "Bread"],
    ["Milk", "Bread", "Butter"],
    ["Bread", "Butter"],
    ["Milk", "Butter"],
    ["Bread", "Butter", "Eggs"],
    ["Milk", "Bread", "Eggs"],
    ["Milk", "Bread", "Butter", "Eggs"],
    ["Bread", "Eggs"],
    ["Milk", "Eggs"],
    ["Milk", "Butter", "Eggs"]
]
```

4. Transaction Encoding

```python
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)

print(df.head())
```

5. Apply Apriori Algorithm

```python
frequent_itemsets = apriori(df, min_support=0.3, use_colnames=True)
frequent_itemsets
```

6. Generate Association Rules

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
rules = rules.sort_values(by="confidence", ascending=False)
rules
```

7. Output Frequent Itemsets and Rules

```python
print("Frequent Itemsets:")
print(frequent_itemsets)

print("\nAssociation Rules:")
print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```

8. Interpretation

Frequent itemsets: Groups of items that occur together in transactions above the support threshold.

Association rules: Imply relationships between items, e.g., {Milk} → {Bread} means if a customer buys Milk, they are likely to buy Bread.
