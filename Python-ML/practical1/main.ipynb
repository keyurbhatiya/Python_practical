# Titanic Data Preprocessing

In this notebook, we will perform step-by-step preprocessing on the Titanic dataset:

1. Handling Missing Values  
2. Encoding Categorical Variables  
3. Scaling Numerical Features  
4. Creating New Features  


## Step 1: Import Required Libraries


```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

## Step 2: Load the Dataset


```python
df = pd.read_csv("titanic.csv")
df
```

## Step 3: Handling Missing Values

- **Age & Fare** → Fill with median (robust against outliers).  
- **Embarked** → Fill with most frequent value (mode).  


```python
# Check missing values
df.isnull().sum()
print("output \n",df)

# Impute median for numerical columns
imputer_median = SimpleImputer(strategy="median")
df[["Age", "Fare"]] = imputer_median.fit_transform(df[["Age", "Fare"]])

# Impute mode for categorical column
imputer_mode = SimpleImputer(strategy="most_frequent")
df[["Embarked"]] = imputer_mode.fit_transform(df[["Embarked"]])

df.isnull().sum()
```

## Step 4: Encoding Categorical Variables

- **Gender** → Binary encoding (Male=0, Female=1).  
- **Embarked** → One-hot encoding for multiple categories.  


```python
# Encode Gender
df["Gender"] = df["Gender"].map({"Male": 0, "Female": 1})

# One-Hot Encode Embarked
df = pd.get_dummies(df, columns=["Embarked"], drop_first=True)

df.head(10)
```

## Step 5: Scaling Numerical Features

We scale the numerical features for better performance in ML models:
- Age  
- Fare  
- Siblings_Spouses  


```python
scaler = StandardScaler()
num_features = ["Age", "Fare", "Siblings_Spouses"]

df[num_features] = scaler.fit_transform(df[num_features])
df.head(10)
```

## Step 6: Creating New Features

We engineer new features to add more information:
- **Family_Size** = Siblings/Spouses + Self  
- **Age_Group** = Age bins (Child, Young, Adult, Senior)  


```python
# Family size feature
df["Family_Size"] = df["Siblings_Spouses"] + 1

# Age group feature
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[-np.inf, -0.5, 0.5, 1.5, np.inf],
    labels=["Child", "Young", "Adult", "Senior"]
)

df.head(10)
```

## Step 7: Compare Original Data vs Final Preprocessed Output


```python
# Reload original dataset to compare
df_original = pd.read_csv("titanic.csv")

print("🔹 Original Titanic Data (with NaN values):")
display(df_original.head(10))

print("\n✅ Final Titanic Data Preprocessing Completed:")
display(df.head(10))
```

## ✅ Final Output


```python
print("Titanic Data Preprocessing Completed ✅")
df.head(10)
```

# 📌 Summary
- Missing values handled with Median & Mode  
- Categorical variables encoded (Gender + Embarked)  
- Features scaled using StandardScaler  
- New features created: Family_Size, Age_Group
