```python
import math

# use math.factorial to compute the factorial of each number in a list

nums = [1,2,3,4,5]

factorials = [math.factorial(i) for i in nums]
print(factorials)
```

```python
# Alternatively, we can define our own factorial function using list

def fact(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

nums = [1, 2, 3, 4, 5]
factorials = [fact(i) for i in nums]
print(factorials)
    
```
